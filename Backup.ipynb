# **Backing Up Linux Mint Settings**

**________________________________________________________________________________________________________________________________________________________________________________________________**

### **1. For Custom Shortcuts & System Tweaks (The `dconf` method)**

**To Back Up:** `dconf dump / > linux_mint_tweaks.ini`

**To Restore:** `dconf load / < linux_mint_tweaks.ini`

**________________________________________________________________________________________________________________________________________________________________________________________________**

### **2. For Terminal Tools & Installed Programs (Mint Backup)**

    - Linux Mint has a brilliant built-in tool called Backup Tool (search for "Backup Tool" in your Mint menu).
    - Open it and click "Backup Software Selection".
    - It will scan your system and create a small text file listing every terminal tool and app you have installed.
    - On your fresh install, open the tool again, click "Restore Software Selection", select that file, and Linux Mint will automatically reinstall every single tool you had via the terminal in one go.
